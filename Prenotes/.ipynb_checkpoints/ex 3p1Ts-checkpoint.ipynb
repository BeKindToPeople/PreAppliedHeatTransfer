```python
import numpy as np
import matplotlib.pyplot as plt

# ==========================================
# 1. DEFINE YOUR PARAMETERS
# ==========================================
# Geometry & Properties
L_sf   = 0.005  # Thickness of skin/fat layer [m] (e.g. 5mm)
k_sf   = 0.37   # Thermal conductivity of skin [W/m*K]
q_dot  = 5000   # Metabolic heat generation [W/m^3]

```

```python
# Boundary Conditions
T_core = 37.0   # Inner body temperature [C]
T_inf  = 20.0   # Ambient fluid temperature [C]

# External Resistance (Insulation + Convection)
# You can calculate R_tot separately if you have k_ins and h
L_ins  = 0.01   # Insulation thickness [m]
k_ins  = 0.04   # Insulation conductivity [W/m*K]
h      = 10     # Convection coefficient [W/m^2*K]

R_tot  = (L_ins / k_ins) + (1 / h)

```

```python
# ==========================================
# 2. CALCULATE CONSTANTS (The "Handshake")
# ==========================================
# C1 is easy (Boundary Condition at x=0)
C1 = T_core

# C0 is the complex algebraic solution we derived
# Numerator terms
term1 = R_tot * q_dot * L_sf
term2 = (q_dot / (2 * k_sf)) * L_sf**2
term3 = T_inf - T_core

# Denominator terms
denom = (L_sf / k_sf) + R_tot

# Calculate C0
C0 = (term1 + term2 + term3) / denom

```

```python
print(f"Calculated C0 (Flux Constant): {C0:.4f}")
print(f"Calculated C1 (Core Temp): {C1:.4f}")

# ==========================================
# 3. CALCULATE TEMPERATURE DISTRIBUTION
# ==========================================
# Create an array of x positions from 0 to L_sf
x = np.linspace(0, L_sf, 100)

# The Temperature Equation: T(x) = -q_dot/2k * x^2 + (C0/k)*x + C1
T_dist = -(q_dot / (2 * k_sf)) * x**2 + (C0 / k_sf) * x + C1

# Calculate the specific surface temperature at x = L_sf
T_surface = T_dist[-1]
print(f"Surface Temperature (Skin/Insulation Interface): {T_surface:.2f} C")

```

```python
# ==========================================
# 4. PLOT THE RESULTS
# ==========================================
plt.figure(figsize=(8, 5))
plt.plot(x * 1000, T_dist, label='Skin/Fat Layer', color='red', linewidth=2)

# Formatting
plt.xlabel('Position x (mm)')
plt.ylabel('Temperature (°C)')
plt.title(f'Temperature Distribution in Tissue (Ts = {T_surface:.1f}°C)')
plt.grid(True, linestyle='--', alpha=0.6)
plt.axhline(T_core, color='gray', linestyle=':', label='Core Temp')
plt.legend()

plt.show()
```
